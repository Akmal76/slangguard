# GPU check

Confirms that this Kaggle notebook can use a GPU before any training work.

On Kaggle: **Settings → Accelerator → GPU T4 x2** (or P100), then **Run All**.

```python
import sys

import torch

print(f"python: {sys.version.split()[0]}")
print(f"torch:  {torch.__version__}")
print(f"cuda available: {torch.cuda.is_available()}")
```

```python
if not torch.cuda.is_available():
    raise RuntimeError("No GPU found. Turn on a GPU accelerator in the notebook settings.")

for i in range(torch.cuda.device_count()):
    print(f"gpu {i}: {torch.cuda.get_device_name(i)}")
```

```python
x = torch.randn(1024, 1024, device="cuda")
y = x @ x
torch.cuda.synchronize()
print(f"matmul ok on {y.device}, result shape {tuple(y.shape)}")
```
